```
# ✅ Install required libraries
!pip install transformers --quiet
!pip install sentencepiece --quiet

# ✅ Import the summarization pipeline from Hugging Face Transformers
from transformers import pipeline

# ✅ Load the model
summarizer = pipeline("summarization")

# ✅ Input: You can replace this text with any lengthy article or paragraph
text = """
India is a country with a rich history, diverse culture, and a rapidly growing economy.
Over the past few decades, India has made significant progress in technology, education,
and infrastructure. However, it also faces challenges such as poverty, pollution, and
political instability. To ensure sustainable development, the government and citizens
must work together to create inclusive policies, invest in renewable energy, and
prioritize education and healthcare for all.
"""

# ✅ Print the original input text
print("Original Text:\n")
print(text)

# ✅ Generate and display the summary
summary = summarizer(text, max_length=80, min_length=30, do_sample=False)
print("\nSummarized Text:\n")
print(summary[0]['summary_text'])
```
