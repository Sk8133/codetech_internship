```python
# Install required libraries
!pip install transformers --quiet
!pip install sentencepiece --quiet

# Import necessary libraries
from transformers import pipeline

# Load the summarization pipeline
summarizer = pipeline("summarization")

# Input Text (you can replace this with any lengthy article)
text = """
The global rise in temperature over the past century has been a major concern for scientists and policymakers.
Climate change, driven primarily by greenhouse gas emissions from human activities, has led to increased
frequency of extreme weather events such as hurricanes, droughts, and wildfires. Governments worldwide are
being urged to take stronger actions to reduce carbon emissions and promote renewable energy.
Meanwhile, individuals can also contribute by reducing energy consumption, using public transport, and supporting
eco-friendly initiatives. The transition to a sustainable future requires cooperation at all levels, from global
leaders to local communities.
"""

# Display the original text
print("Original Text:\n", text)

# Perform summarization
summary = summarizer(text, max_length=80, min_length=30, do_sample=False)

# Display the summary
print("\nSummarized Text:\n", summary[0]['summary_text'])
```
